```python
marks = [85, 90, 78, 92, 88]

# Find highest mark manually
highest = marks[0]
for mark in marks:
    if mark > highest:
        highest = mark

# Find lowest mark manually
lowest = marks[0]
for mark in marks:
    if mark < lowest:
        lowest = mark

# Find average manually
total = 0
count = 0
for mark in marks:
    total += mark
    count += 1

average = total / count

print("Highest Mark:", highest)
print("Lowest Mark:", lowest)
print("Average Mark:", average)
```

```python
# Small integers — Python CACHES them, so same object
a = 10
c = 10
print(a == c)    # True  (same value)
print(a is c)    # True  (same object — Python reuses small ints)

# Lists — same content, DIFFERENT objects
lst1 = [1, 2, 3]
lst2 = [1, 2, 3]
print(lst1 == lst2)   # True  (same content)
print(lst1 is lst2)   # False (different objects in memory!)
```

```python
# Python way — no temp variable needed!
a = 10
b = 20
a, b = b, a        # swap in one line
print(a, b)        # 20 10

# Traditional way (with temp variable)
a = 10
b = 20
temp = a
a = b
b = temp
print(a, b)        # 20 10
```

```python
word = "aeroplane"
word_count = {}

for char in word:
    
    if char in word_count:
        word_count[char] += 1
    else:
        word_count[char] = 1
        
print(word_count)
```
