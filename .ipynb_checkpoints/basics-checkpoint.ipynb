```python
drama_club = {"alice","bob","charlie","david","eve"}
music_club = {"charlie","david","frank","grace",}
sports_club = {"alice","eve","frank","henry"}
```

```python
all_drama_music = drama_club.union(music_club)
print("all drama/music students:", all_drama_music)
```

```python
# 2. INTERSECTION (&) - Students in BOTH drama AND music
both_clubs = drama_club.intersection(music_club)
print("In both drama & music:", both_clubs)
```

```python
# 3. DIFFERENCE (-) - Students in drama but NOT in music
only_drama = drama_club.difference(music_club)
print("Only in drama:", only_drama)
```

```python
# 4. SYMMETRIC DIFFERENCE (^) - Students in either but NOT both
exclusive = drama_club.symmetric_difference(music_club)
print("In one club only:", exclusive)
# {'Alice', 'Bob', 'Eve', 'Frank', 'Grace'}
```

```python
# 5. SUBSET - Is music_club a subset of drama_club?
print("Music ⊆ Drama?", music_club.issubset(drama_club)) 
```

```python
drama_club = {"Alice", "Bob", "Charlie", "David", "Eve"}
small = {"Alice", "Bob"}
print("{Alice,Bob} ⊆ Drama?", small.issubset(drama_club))
```

```python
# 6. SUPERSET - Does drama_club contain all of small?
print("Drama ⊇ {Alice,Bob}?", drama_club.issuperset(small)) 
```

```python
# 7. DISJOINT - Do music and sports share NO members?
print("Music & Sports disjoint?", music_club.isdisjoint(sports_club))
```

```python
drama_club.add("Zara")           # Add one element
drama_club.update({"Ian", "Jo"}) # Add multiple elements
print("After adding:", drama_club)
```

```python
drama_club.discard("Zara")       # Remove (no error if missing)
drama_club.remove("Jo")         # Remove (raises error if missing)
popped = drama_club.pop()       # Remove & return a random element
print(f"Popped: {popped}, Remaining: {drama_club}")
```

```python
def remove_duplicates(nums):
    
    unique_nums = []
    for num in nums:
        if num not in unique_nums:
            unique_nums.append(num)
            
    return unique_nums
lst = [1,12,23,4,54,2,4,5,6,8,5,6,8,5]
remove_duplicates(lst)
```
