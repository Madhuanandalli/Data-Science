```python
import pandas as pd

df = pd.read_excel("GFC_10K_Financial_Data_2023_2025.xlsx")
df
```

# GFC Financial Data Analysis

## Objective

The objective of this analysis is to examine key financial data from the 10-K filings of Microsoft, Tesla, and Apple for 2023 to 2025.

The analysis focuses on:

- Total Revenue
- Net Income
- Total Assets
- Total Liabilities
- Cash Flow from Operating Activities

The extracted financial data is used to identify year-over-year trends and indicators of financial performance and health.

```python
df.info()
```

```python
df.describe()
```

```python
df = df.sort_values(["Company", "Fiscal Year"])

df["Revenue Growth (%)"] = (
    df.groupby("Company")["Total Revenue"].pct_change() * 100
)

df[["Company", "Fiscal Year", "Total Revenue", "Revenue Growth (%)"]]
```

```python
df["Net Income Growth (%)"] = (
    df.groupby("Company")["Net Income"].pct_change() * 100
)

df[["Company", "Fiscal Year", "Net Income", "Net Income Growth (%)"]]
```

```python
df["Operating Cash Flow Growth (%)"] = (
    df.groupby("Company")["Operating Cash Flow"].pct_change() * 100
)

df[
    [
        "Company",
        "Fiscal Year",
        "Operating Cash Flow",
        "Operating Cash Flow Growth (%)"
    ]
]
```

```python
df["Net Profit Margin (%)"] = (
    df["Net Income"] / df["Total Revenue"] * 100
)
```

```python
df["Liabilities / Assets (%)"] = (
    df["Total Liabilities"] / df["Total Assets"] * 100
)
```

```python
df["Operating Cash Flow / Revenue (%)"] = (
    df["Operating Cash Flow"] / df["Total Revenue"] * 100
)
```

```python
df.round(2)
```

# Initial Findings

## Microsoft

Microsoft shows strong growth in revenue, net income, and operating cash flow between 2023 and 2025. The simultaneous improvement in profitability and operating cash generation indicates strong financial performance over the period.

## Tesla

Tesla's revenue remained relatively flat between 2023 and 2025, while net income declined significantly. This indicates a substantial deterioration in profitability despite relatively stable operating cash flow.

## Apple

Apple experienced revenue growth between 2023 and 2025, accompanied by an increase in net income. Operating cash flow remained strong throughout the period, indicating continued ability to generate cash from operations.

## Cross-company observation

The analysis demonstrates that revenue growth alone does not provide a complete picture of financial health. Profitability, balance-sheet position, and operating cash flow should also be considered.

These findings can inform the future AI-powered financial chatbot by providing structured financial metrics that can be used for quantitative analysis and trend identification.

```python
summary = df.groupby("Company").agg(
    Revenue_2023=("Total Revenue", "first"),
    Revenue_2025=("Total Revenue", "last"),
    Net_Income_2023=("Net Income", "first"),
    Net_Income_2025=("Net Income", "last"),
    OCF_2023=("Operating Cash Flow", "first"),
    OCF_2025=("Operating Cash Flow", "last")
).reset_index()

summary
```

```python
summary["Revenue Change (%)"] = (
    (summary["Revenue_2025"] - summary["Revenue_2023"])
    / summary["Revenue_2023"] * 100
)

summary["Net Income Change (%)"] = (
    (summary["Net_Income_2025"] - summary["Net_Income_2023"])
    / summary["Net_Income_2023"] * 100
)

summary["OCF Change (%)"] = (
    (summary["OCF_2025"] - summary["OCF_2023"])
    / summary["OCF_2023"] * 100
)

summary.round(2)
```

## Recommendation for GFC's AI Financial Chatbot

The chatbot should combine quantitative financial calculations with qualitative explanations from company filings.

For example, instead of simply answering:

"Tesla's net income decreased."

The chatbot should provide:

**Data → Finding → Driver → Implication → Recommendation**

This approach would allow users to move from financial figures to actionable business insights.

```python
df.head()
```

```python
def simple_chatbot(user_query):
    user_query = user_query.strip().lower()

    # Query 1: Microsoft's revenue in 2025
    if user_query == "what was microsoft's revenue in 2025?":
        revenue = df[(df["Company"] == "Microsoft") & (df["Fiscal Year"] == 2025)]["Total Revenue"].iloc[0]
        return f"Microsoft's revenue in 2025 was ${revenue:,.0f} million."

    # Query 2: Apple's net income in 2025
    elif user_query == "what was apple's net income in 2025?":
        net_income = df[(df["Company"] == "Apple") & (df["Fiscal Year"] == 2025)]["Net Income"].iloc[0]
        return f"Apple's net income in 2025 was ${net_income:,.0f} million."

    # Query 3: Tesla's operating cash flow in 2025
    elif user_query == "what was tesla's operating cash flow in 2025?":
        cash_flow = df[(df["Company"] == "Tesla") & (df["Fiscal Year"] == 2025)]["Operating Cash Flow"].iloc[0]
        return f"Tesla's operating cash flow in 2025 was ${cash_flow:,.0f} million."

    # Query 4: Microsoft's revenue growth in 2025
    elif user_query == "what was microsoft's revenue growth in 2025?":
        growth = df[(df["Company"] == "Microsoft") & (df["Fiscal Year"] == 2025)]["Revenue Growth (%)"].iloc[0]
        return f"Microsoft's revenue grew by {growth:.2f}% from 2024 to 2025."

    # Query 5: Company with highest net profit margin in 2025
    elif user_query == "which company had the highest net profit margin in 2025?":
        data_2025 = df[df["Fiscal Year"] == 2025]
        highest = data_2025.loc[data_2025["Net Profit Margin (%)"].idxmax()]
        return f"{highest['Company']} had the highest net profit margin in 2025 at {highest['Net Profit Margin (%)']:.2f}%."

    # Error handling
    else:
        return (
            "Sorry, I can only provide information on predefined financial queries. "
            "You can ask about revenue, net income, operating cash flow, "
            "revenue growth, or net profit margin."
        )
```

```python
simple_chatbot("what was microsoft's revenue in 2025?")
```

```python
print(simple_chatbot("what was apple's net income in 2025?"))

print(simple_chatbot("what was tesla's operating cash flow in 2025?"))

print(simple_chatbot("what was microsoft's revenue growth in 2025?"))

print(simple_chatbot("which company had the highest net profit margin in 2025?"))
```

```python
simple_chatbot("what is Apple's stock price tomorrow?")
```

```python
print("Welcome to the GFC Financial Chatbot!")
print("Ask me a predefined financial question.")
print("Type 'exit' to end the conversation.\n")

while True:
    user_query = input("You: ")

    if user_query.lower().strip() == "exit":
        print("Chatbot: Thank you for using the GFC Financial Chatbot. Goodbye!")
        break

    response = simple_chatbot(user_query)
    print("Chatbot:", response)
```

```python
# Test Results

test_queries = [
    "what was microsoft's revenue in 2025?",
    "what was apple's net income in 2025?",
    "what was tesla's operating cash flow in 2025?",
    "what was microsoft's revenue growth in 2025?",
    "which company had the highest net profit margin in 2025?",
    "what is Apple's stock price tomorrow?"
]

for query in test_queries:
    print("User:", query)
    print("Chatbot:", simple_chatbot(query))
    print("-" * 80)
```

# GFC Financial Chatbot – Prototype Documentation

## 1. Overview

The GFC Financial Chatbot is a rule-based prototype designed to provide quick answers to predefined financial queries using the financial data analyzed in Task 1.

The chatbot uses Python and pandas to retrieve financial information from the structured dataset and provide simple, user-friendly responses.

## 2. How the Chatbot Works

The chatbot follows a simple rule-based approach using `if`, `elif`, and `else` statements.

1. The user enters a financial question.
2. The chatbot converts the input to lowercase and removes unnecessary spaces.
3. The chatbot compares the question with predefined queries.
4. If a matching query is found, the chatbot retrieves the relevant financial data from the dataset.
5. The chatbot returns the result in a readable format.
6. If the question is not supported, the chatbot provides a message explaining the types of questions it can answer.
7. The chatbot can be exited by entering `exit`.

## 3. Predefined Queries

The chatbot currently supports the following queries:

1. What was Microsoft's revenue in 2025?
2. What was Apple's net income in 2025?
3. What was Tesla's operating cash flow in 2025?
4. What was Microsoft's revenue growth in 2025?
5. Which company had the highest net profit margin in 2025?

## 4. Testing

The chatbot was tested using all five predefined queries and an unsupported financial question.

All five predefined queries returned the expected financial information. The unsupported question was handled gracefully using the chatbot's fallback response.

## 5. Limitations

This prototype has several limitations:

* It only responds to predefined questions.
* It does not use advanced natural language processing.
* Small changes in the wording of a question may result in an unsupported-query response.
* It does not learn from previous conversations.
* It does not provide real-time financial or stock-market information.
* It does not include a web-based user interface.
* The chatbot is limited to the financial data available in the Task 1 dataset.

## 6. Conclusion

This prototype demonstrates how rule-based logic can be used to create a simple financial chatbot. It successfully connects structured financial data with predefined user queries and provides reliable responses within its defined scope.
